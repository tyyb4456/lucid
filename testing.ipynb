```python
import os
from typing import Literal

from dotenv import load_dotenv
load_dotenv()
```

```python
from deepagents import create_deep_agent

agent = create_deep_agent(
    model="google_genai:gemini-2.5-flash-lite",
    system_prompt="you are an assistant",
)
```

```python
result = agent.invoke({"messages": [{"role": "user", "content": "What is ai -- tell me in one line"}]})

# Print the agent's response
print(result["messages"][-1].content)
```

```python
def internet_search():
    """Run a web search"""
    return "i found something awesome on internet"


research_subagent = {
    "name": "research-agent",
    "description": "Used to research more in depth questions",
    "system_prompt": "You are a great researcher",
    "tools": [internet_search],
}


subagents = [research_subagent]
```

```python
from deepagents import create_deep_agent

agent = create_deep_agent(
    model="google_genai:gemini-2.5-flash-lite",
    system_prompt="you are an assistant, that use subagent",
    subagents = subagents
)
```

```python
result = agent.invoke({"messages": [{"role": "user", "content": "What is ai -- search on inetrent"}]})

# Print the agent's response
print(result["messages"][-1].content)
```
